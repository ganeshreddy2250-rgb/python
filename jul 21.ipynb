```python
n = int(input())

count = 0

while count < 3:
    pin = int(input())

    if pin == n:
        print("Access Granted")
        break
    else:
        print("Wrong PIN,", 2 - count, "tries left")

    count = count + 1

if count == 3:
    print("Card Locked")
```

```python
n = int(input())

for i in range(3):
    pin = int(input())

    if pin == n:
        print("Access Granted")
        break
    else:
        print("Wrong PIN,", 2 - i, "tries left")

if pin != n:
    print("Card Locked")
```

```python
''' Grocery Checkout Total'''
```

```python
item_prices = [200, 300, 600, 400, 100]

total = 0
i = 0
discount = False

while i < len(item_prices):

    if discount:
        total = total + (item_prices[i] * 0.9)
    else:
        total = total + item_prices[i]

    if total > 1000:
        discount = True

    i = i + 1

print("Final Total:", total)
```

```python
item_prices = [200, 300, 600, 400, 100]

total = 0

for price in item_prices:
    if total > 1000:
        total = total + (price * 0.9)
    else:
        total = total + price

print("Final Total:", total)
```

```python
''' Number Guessing Game'''
```

```python
secret_number = 25
guesses = [10, 30, 20, 25]

count = 0

for guess in guesses:
    count = count + 1

    if guess == secret_number:
        print("Correct")
        print("Total guesses:", count)
        break

    elif guess > secret_number:
        print("Too High")

    else:
        print("Too Low")
```

```python
secret_number = 25
guesses = [10, 30, 20, 25]

i = 0
count = 0

while i < len(guesses):
    count = count + 1

    if guesses[i] == secret_number:
        print("Correct")
        print("Total guesses:", count)
        break

    elif guesses[i] > secret_number:
        print("Too High")

    else:
        print("Too Low")

    i = i + 1
```

```python
total_seats = 20
booking_requests = [5, 4, 6, 8]

i = 0

while i < len(booking_requests):

    if booking_requests[i] > total_seats:
        print("Not enough seats")
        break

    total_seats = total_seats - booking_requests[i]
    i = i + 1

print("Seats remaining:", total_seats)
```

```python
total_seats = 20
booking_requests = [5, 4, 6, 8]

for seats in booking_requests:

    if seats > total_seats:
        print("Not enough seats")
        break

    total_seats = total_seats - seats

print("Seats remaining:", total_seats)
```

```python
'''Study Streak Tracker'''
```

```python
scores = [80, 75, 90, 55, 70]

streak = 0

for score in scores:

    if score >= 60:
        streak = streak + 1

    else:
        print("Streak broken")
        break

print("Final streak:", streak)
```
