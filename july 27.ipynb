```python
class Student:

    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"Student Name: {self.name}"

s = Student("Sai")

print(s)
```

```python
class Student:

    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"Name: {self.name}"

    def __repr__(self):
        return f"Student('{self.name}')"

s = Student("Sai")

print(s)
print(repr(s))
```

```python
class Student:

    def __init__(self):
        self.name = "Sai"

s = Student()

s.__dict__["age"] = 24

print(s.__dict__)
```

```python
class Student:

    # 1. Constructor
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # 2. User-friendly representation
    def __str__(self):
        return f"Student Name: {self.name}, Age: {self.age}"

    # 3. Developer representation
    def __repr__(self):
        return f"Student(name='{self.name}', age={self.age})"


# Creating an object
s1 = Student("Sai", 24)

# __str__() is called automatically
print(s1)

# __repr__() is called
print(repr(s1))

# __dict__ returns all instance variables
print(s1.__dict__)
```
